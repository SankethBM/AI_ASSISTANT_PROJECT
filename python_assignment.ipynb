17. Write a program that checks if a point(x,y) lies inside, outside  or on a circle with a given center (cx,cy) and radius.

```python
import math

x = float(input("Enter x-coordinate of the point: "))
y = float(input("Enter y-coordinate of the point: "))
cx = float(input("Enter x-coordinate of the circle center: "))
cy = float(input("Enter y-coordinate of the circle center: "))
r = float(input("Enter radius of the circle: "))

distance = math.sqrt((x - cx)**2 + (y - cy)**2)

if distance < r:
    print("🟢 The point lies inside the circle.")
elif distance == r:
    print("🟡 The point lies on the circle.")
else:
    print("🔴 The point lies outside the circle.")
```

19. Write a python program to illustrate ATM Transaction.

```python
user_pin = "1234"
balance = 10000.0

print("🏦 Welcome to Simple ATM")

entered_pin = input("🔐 Enter your 4-digit PIN: ")

if entered_pin == user_pin:
    print("✅ Access Granted\n")

    while True:
        print("\nChoose a transaction:")
        print("1. 💰 Check Balance")
        print("2. ➕ Deposit")
        print("3. ➖ Withdraw")
        print("4. 🚪 Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            print(f"💳 Your current balance is: Rs {balance:.2f}")

        elif choice == "2":
            amount = float(input("Enter amount to deposit: Rs "))
            if amount > 0:
                balance += amount
                print(f"✅ Rs {amount:.2f} deposited successfully.")
            else:
                print("❌ Invalid deposit amount.")

        elif choice == "3":
            amount = float(input("Enter amount to withdraw: Rs "))
            if 0 < amount <= balance:
                balance -= amount
                print(f"✅ Rs {amount:.2f} withdrawn successfully.")
            elif amount > balance:
                print("❌ Insufficient balance.")
            else:
                print("❌ Invalid withdrawal amount.")

        elif choice == "4":
            print("🙏 Thank you for using the ATM. Goodbye!")
            break

        else:
            print("❌ Invalid option. Please choose 1-4.")
else:
    print("❌ Invalid PIN. Access Denied.")
```

```python
def v_c(str):
    vowels = 'aeiou'
    c = 0
    for char in str.lower():
        if char in vowels:
            c += 1
    return c

str = input("Enter a string: ")

print(f"Number of vowels in the string: {v_c(str)}")
```

4.Write a program that takes a string and prints the frequency of each character in it.

```python
str = input("Enter a string: ")
freq = {}

for c in str:
    if c in freq:
        freq[c] += 1
    else:
        freq[c] = 1

print("Character Frequency:")
for c, count in freq.items():
  print(f"{c}: {count}")
```

2.Write the function which performs following strings operation (Not to Use Built In
Methods)

a. Case conversion
b. Count the numbers character
c. Count the specific character count
d. Reverse the string
e. Remove all white spaces in the string

```python

def case_conversion(s):
    result = ""
    for ch in s:
        if 'a' <= ch <= 'z':
            result += chr(ord(ch) - 32)
        elif 'A' <= ch <= 'Z':
            result += chr(ord(ch) + 32)
        else:
            result += ch
    return result

def count_characters(s):
    count = 0
    for _ in s:
        count += 1
    return count

def count_specific(s, target):
    cnt = 0
    for ch in s:
        if ch == target:
            cnt += 1
    return cnt

def reverse_string(s):
    rev = ""
    for ch in s:
        rev = ch + rev
    return rev

def remove_spaces(s):
    result = ""
    for ch in s:
        if ch != ' ':
            result += ch
    return result


text = "Hello World 123"

print("Original:", text)
print("Case Conversion:", case_conversion(text))
print("Character Count:", count_characters(text))
print("Count of 'l':", count_specific(text, 'l'))
print("Reversed:", reverse_string(text))
print("Without Spaces:", remove_spaces(text))
```

4. Write a program that stores city names along with their latitude and longitude as
tuples. The program should:
a. Store multiple cities with their geographic coordinates.
b. Accept input of two city names from the user.
c. Calculate and display the approximate distance between the two cities using
the Haversine formula or Euclidean distance.

```python
import math

cities = {
    "Bengaluru": (12.9716, 77.5946),
    "Mysuru": (12.2958, 76.6394),
    "Hubballi": (15.3647, 75.1230),
    "Belagavi": (15.8497, 74.4977),
    "Mangaluru": (12.9141, 74.8560)
}

def haversine(c1, c2):
    R = 6371
    lat1, lon1 = map(math.radians, cities[c1])
    lat2, lon2 = map(math.radians, cities[c2])
    dlat, dlon = lat2 - lat1, lon2 - lon1
    a = math.sin(dlat/2)**2 + math.cos(lat1)*math.cos(lat2)*math.sin(dlon/2)**2
    return R * 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

c1, c2 = input("Enter two cities (comma separated): ").split(",")
c1, c2 = c1.strip(), c2.strip()

if c1 in cities and c2 in cities:
    print(f"Distance: {haversine(c1, c2):.2f} km")
else:
    print("City not found.")
```

6. Store execution times of different sorting algorithms on various input sizes as tuples.
Analyze and display which algorithm performs best under given conditions.

```python
import time, random

def bubble(arr):
    for i in range(len(arr)-1):
        for j in range(len(arr)-1-i):
            if arr[j] > arr[j+1]: arr[j], arr[j+1] = arr[j+1], arr[j]

def insertion(arr):
    for i in range(1, len(arr)):
        key = arr[i]; j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j+1] = arr[j]; j -= 1
        arr[j+1] = key

def quick(arr):
    if len(arr) <= 1: return arr
    p = arr[len(arr)//2]
    return quick([x for x in arr if x < p]) + [x for x in arr if x == p] + quick([x for x in arr if x > p])

sizes = [100, 500, 1000]
algos = [("Bubble", bubble), ("Insertion", insertion), ("Quick", lambda a: quick(a))]

results = []
for s in sizes:
    data = [random.randint(1, 10000) for _ in range(s)]
    for name, algo in algos:
        arr = data[:]
        start = time.time()
        algo(arr)
        results.append((name, s, round(time.time() - start, 5)))

print("Algorithm Times:")
for a, sz, t in results:
    print(f"{a:9} Size={sz:<5} Time={t}")

for s in sizes:
    best = min([(a, t) for a, sz, t in results if sz == s], key=lambda x: x[1])
    print(f"Best for size {s}: {best[0]} ({best[1]}s)")
```

7. Implement Fibonacci number generation using memoization. Store computed
Fibonacci values in a tuple or immutable structure to avoid recomputation.

```python
def fib(n, m=(0, 1)):
    if n < len(m):
        return m[n], m
    else:
        f1, m = fib(n - 1, m)
        f2, m = fib(n - 2, m)
        m = m + (f1 + f2,)
        return m[-1], m

n = int(input("Enter n: "))
result, _ = fib(n)
print(f"Fibonacci number F({n}) = {result}")
```

```python
from collections import Counter
import re

sample_text = """
Python is a powerful programming language.
Python is easy to learn and fun to use.
Many developers enjoy Python for data analysis, web development, and automation.
"""

with open("article.txt", "w") as f:
    f.write(sample_text)

print("article.txt file created.\n")

with open("article.txt", "r") as f:
    text = f.read().lower()

words = re.findall(r"\b[a-zA-Z]+\b", text)
word_count = Counter(words)

print("Word Frequencies:\n")
for word, freq in word_count.items():
    print(word, ":", freq)
```

1. Create a Python program using a class BankAccount.
Allow the user to deposit, withdraw, and check balance. If withdrawal exceeds
balance, display &quot;Insufficient Funds&quot;.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"₹{amount} deposited. Current Balance: ₹{self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient Funds")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.balance -= amount
            print(f"₹{amount} withdrawn. Current Balance: ₹{self.balance}")

    def check_balance(self):
        print(f"Available Balance: ₹{self.balance}")


account = BankAccount("Sanketh")

while True:
    print("\n1. Deposit\n2. Withdraw\n3. Check Balance\n4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        amt = float(input("Enter deposit amount: "))
        account.deposit(amt)
    elif choice == "2":
        amt = float(input("Enter withdrawal amount: "))
        account.withdraw(amt)
    elif choice == "3":
        account.check_balance()
    elif choice == "4":
        print("Exiting... Thank you!")
        break
    else:
        print("Invalid option. Try again.")
```

displaying the final order bill with customer details.

```python
class Order:
    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.items = {}

    def add_item(self, item, price, qty):
        if item in self.items:
            self.items[item][1] += qty
        else:
            self.items[item] = [price, qty]

    def remove_item(self, item):
        if item in self.items:
            del self.items[item]

    def display_bill(self):
        total = 0
        print("\nCustomer:", self.customer_name)
        print("Items Purchased:")
        for item, (price, qty) in self.items.items():
            cost = price * qty
            total += cost
            print(f"{item} - {qty} x ₹{price} = ₹{cost}")
        print("Total Bill: ₹", total)


name = input("Enter customer name: ")
o = Order(name)

while True:
    print("\n1. Add Item\n2. Remove Item\n3. Show Bill\n4. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        item = input("Enter item name: ")
        price = float(input("Enter price: "))
        qty = int(input("Enter qty: "))
        o.add_item(item, price, qty)
    elif choice == "2":
        item = input("Enter item name to remove: ")
        o.remove_item(item)
    elif choice == "3":
        o.display_bill()
    elif choice == "4":
        break
    else:
        print("Invalid choice")
```

3. Write a Python program to demonstrate polymorphism with a method area() defined
in base class Shape and overridden in Circle and Rectangle.

```python
import math

class Shape:
    def area(self):
        return 0

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius * self.radius

class Rect(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = float(input("Enter radius of circle: "))
circle = Circle(r)

l = float(input("Enter len of rect: "))
w = float(input("Enter wid of rect: "))
rect = Rect(l, w)

print("Area of Circle:", circle.area())
print("Area of Rect:", rect.area())
```

9. Create a Python program to demonstrate operator overloading in an E-commerce
system:
 Define a class Cart where the + operator is overloaded to combine two carts
(items).

```python
class Cart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def __add__(self, other):
        new_cart = Cart()
        new_cart.items = self.items + other.items
        return new_cart

    def display_cart(self):
        if not self.items:
            print("Cart is empty.")
        else:
            print("Cart Items:", ", ".join(self.items))


cart1 = Cart()
cart2 = Cart()

n1 = int(input("Enter number of items for Cart 1: "))
for i in range(n1):
    item = input(f"Enter item {i+1} for Cart 1: ")
    cart1.add_item(item)

n2 = int(input("Enter number of items for Cart 2: "))
for i in range(n2):
    item = input(f"Enter item {i+1} for Cart 2: ")
    cart2.add_item(item)

combined_cart = cart1 + cart2

print("\nCart 1:")
cart1.display_cart()
print("\nCart 2:")
cart2.display_cart()
print("\nCombined Cart:")
combined_cart.display_cart()
```

```python
class Device:
    def status(self):
        print("Device status: Unknown")

class Mobile(Device):
    def status(self):
        print("Mobile status: Active")

class Tablet(Device):
    def status(self):
        print("Tablet status: Active")

class Smartphone(Mobile, Tablet):
    pass

phone = Smartphone()
phone.status()

print("\nMethod Resolution Order (MRO):")
for cls in Smartphone.__mro__:
    print(cls.__name__)
```

operator to add balances of two wallets, and the - operator to transfer money between
wallets.

```python
class Wallet:
    def __init__(self, balance=0):
        self.balance = balance

    def __add__(self, other):
        return Wallet(self.balance + other.balance)

    def __sub__(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
            return Wallet(self.balance)
        else:
            self.balance -= amount
            return Wallet(self.balance)

    def display_balance(self):
        print(f"Balance: ₹{self.balance}")


balance1 = float(input("Enter balance for Wallet 1: "))
balance2 = float(input("Enter balance for Wallet 2: "))

w1 = Wallet(balance1)
w2 = Wallet(balance2)

print("\nWallet 1:")
w1.display_balance()
print("Wallet 2:")
w2.display_balance()

combined_wallet = w1 + w2
print("\nCombined Wallet:")
combined_wallet.display_balance()

transfer_amount = float(input("\nEnter amount to transfer from Wallet 1: "))
w1 - transfer_amount
print("Wallet 1 after transfer:")
w1.display_balance()
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)
traffic = np.random.randint(50, 500, size=(24, 6))

hours = [f"{h}:00" for h in range(24)]
junctions = [f"Junction {i+1}" for i in range(6)]

df = pd.DataFrame(traffic, index=hours, columns=junctions)

df["Total"] = df.sum(axis=1)
peak_hours = df["Total"].sort_values(ascending=False).head(5)

junction_totals = df[junctions].sum().sort_values(ascending=False)

traffic_variance = df["Total"].var()

print("Traffic Data (24 hours × 6 junctions):\n", df, "\n")
print("Peak Traffic Hours:\n", peak_hours, "\n")
print("Traffic Load Comparison across Junctions:\n", junction_totals, "\n")
print("Traffic Variance across Hours:", round(traffic_variance, 2))

if traffic_variance > 20000:
    print("Traffic congestion is concentrated in specific time slots.")
else:
    print("Traffic is relatively evenly distributed across the day.")
```
